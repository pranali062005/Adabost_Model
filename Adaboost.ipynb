```python
import pandas as pd
import numpy as np

from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.preprocessing import LabelEncoder
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sharmajicoder/genn-z-college-admission-dataset")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv(f'{path}/genz_college_admission_prediction.csv')
```

```python
df = df.drop(columns='student_id')
```

```python
df.head()
```

```python
df.dtypes
```

```python
OBJ = LabelEncoder()


df['gender'] = OBJ.fit_transform(df['gender'])

df['state'] = OBJ.fit_transform(df['state'])
```

```python
df.dtypes
```

```python
df.columns
```

```python
x = df[['age', 'gender', 'state', 'family_income', 'high_school_gpa',
       'sat_score', 'act_score', 'attendance_rate', 'ap_courses',
       'extracurricular_count', 'volunteer_hours', 'leadership_positions',
       'coding_projects', 'social_media_hours', 'online_certifications',
       'essay_score', 'recommendation_score', 'interview_score']]

y = df['admission_status']
```

```python
x_train , x_test ,y_train , y_test = train_test_split(x,y,test_size = 0.2 , random_state = 42)
```

```python
model = AdaBoostClassifier()
model.fit(x_train,y_train)
```

```python
predict = model.predict(x_test)
predict
```

```python
from sklearn.metrics import accuracy_score

score = accuracy_score(y_test, predict)

print('Accuracy', score)
```

```python
import pickle

pickle.dump(model, open('Adaboost_Model.pkl', 'wb'))
print('PKL File Created Successfully')
```
